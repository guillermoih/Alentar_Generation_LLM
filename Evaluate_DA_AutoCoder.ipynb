```python
import pandas as pd
```

```python
df_da = pd.read_csv('llm_responses_deepseek-r1.csv', delimiter='|')
```

```python
df_da = df_da.drop(['model', 'prompt', 'response_raw', 'time'], axis=1)
df_da = df_da.reset_index(names='id_caso')
df_da = df_da.rename(columns={'response': 'texto_diagnostico'})
```

```python
def normalize_list(value):
    if pd.isna(value) or value == "" or value == "[]":
        return []
    s = str(value)
    
    # Delete [] and ""
    s = s.replace('[', '').replace(']', '').replace('"', '')
    # Split by comma
    s = s.split(',')
    # Clean empty spaces
    s = [item.strip() for item in s if item.strip()]
    
    return s
```

```python
df_da['code'] = df_da['code'].apply(normalize_list)
```

```python
df_da.to_csv('AutoCoder/auto_coder CIE-10/informes_deepseek-r1.csv', index=False)
```
