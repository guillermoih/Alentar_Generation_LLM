```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.alentar_llm.data_load import get_cods_dict, clean_data
```

```python
df = pd.read_csv('Informes-Urgencias/EJEI_nocod_Clasificator_2024_07_ 04. 4.fin.csv')
cods = get_cods_dict('Informes-Urgencias/vars.csv')
df = clean_data(df, cods)
```

```python
text_len = df['EJEIDiagnóstico'].str.len()
fig, ax = plt.subplots(1, 2, figsize=(12,5))
ax[0].boxplot(text_len)
ax[1].hist(text_len, bins=20)
ax[0].set_ylabel("Text lenght")
ax[1].set_ylabel("Text lenght")
plt.show()
print(f"Mean lenght: {np.mean(text_len)}")
print(f"Standard deviation of lenghts: {np.std(text_len)}")
print(f"Minimum lenght: {min(text_len)}")
print(f"Maximum lenght: {max(text_len)}")
print(f"Median lenght: {np.median(text_len)}")
```
